# Coding Attention Mechanism


 ### A simple self-attention mechanism without trainable weights

```python
import torch

inputs = torch.tensor(
  [[0.43, 0.15, 0.89], # Your     (x^1)
   [0.55, 0.87, 0.66], # journey  (x^2)
   [0.57, 0.85, 0.64], # starts   (x^3)
   [0.22, 0.58, 0.33], # with     (x^4)
   [0.77, 0.25, 0.10], # one      (x^5)
   [0.05, 0.80, 0.55]] # step     (x^6)
)
```

```python
inputs_query =inputs[1]
inputs_query
```

```python
input_1 = inputs[0]
input_1
```

```python
torch.dot(input_1,inputs_query)
```

```python
res = 0.

for idx, element in enumerate(inputs[0]):
    res += inputs[0][idx] * inputs_query[idx]

print(res)
print(torch.dot(inputs[0], inputs_query))
```

```python
torch.empty(inputs.shape[0])
```

```python
inputs_query = inputs[1]  # 2nd input token is the query

attn_scores_2 = torch.empty(inputs.shape[0])
for i, x_i in enumerate(inputs):
    attn_scores_2[i] = torch.dot(x_i, inputs_query) # dot product (transpose not necessary here since they are 1-dim vectors)

print(attn_scores_2)
```

```python
# brute force way to normalise is to divide by the total sum
attn_weights_temp = attn_scores_2 / attn_scores_2.sum()
attn_weights_temp
```

```python
attn_weights_temp.sum()
```

```python
def softmax_naive(x):
    return torch.exp(x) / torch.exp(x).sum(dim=0)

attn_weights_2 = softmax_naive(attn_scores_2)
attn_weights_2
```

```python
query = inputs[1] # 2nd input token is the query

context_vec_2 = torch.zeros(query.shape)
for i,x_i in enumerate(inputs):
    # print(attn_weights_2[i]*x_i)
    context_vec_2 += attn_weights_2[i]*x_i


print(context_vec_2)
```
